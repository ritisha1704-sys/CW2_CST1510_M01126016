| Term | Formula |
|------|---------|
| Gross profit | `(sell_price − buy_price) × quantity` |

| Net profit | `gross_profit − fee_amount` |

### What to do

## What you need to know

### Making decisions with `if / elif / else`
Python runs different code depending on whether a condition is `True` or `False`.

```python
score = 72
```

> **Important:** Python uses indentation (4 spaces or 1 tab) to show which lines belong inside the `if` block.

A `for` loop runs a block of code a fixed number of times.

```python

    print(f"Step {i}")
```

```
Step 3
Step 4
```

You can put an `if` statement **inside** a loop to check something on every step.

# Week 3 — Functions

**Concepts:** `def` · parameters · `return` · calling functions · keeping functions focused

---

## What you need to know

### Why use functions?
Without functions, if you want to do the same calculation for 10 different trades you have to copy-paste the same lines 10 times.  
A **function** is a named block of code you define once and call as many times as you need.

### Defining a function

```python
def function_name(parameter1, parameter2):
    # code goes here
    return result
```

- **`def`** — tells Python you are defining a function.
- **Parameters** — the input values the function needs (listed in brackets).
- **`return`** — sends a value back to whoever called the function.  
  A function that only prints and never returns is limited — prefer returning and letting the caller decide what to print.

### Calling a function

```python
result = function_name(value1, value2)
print(result)
```

### A function should do one thing
`calculate_profit()` should calculate and return — it should not also ask for input or print anything.  
Keeping functions focused makes them easy to reuse and test.

---

## Worked Example — Unit Converter

```python
def km_to_miles(km):

    """Return the distance in kilometres."""

# calling the functions
```

# Week 4 — Strings and Error Handling

**Concepts:** string methods · `.split()` · `.strip()` · `.upper()` · `try / except` · `ValueError` · `IndexError`

---

## What you need to know

### Useful string methods

| Method | What it does | Example |
|--------|-------------|---------|
| `.split()` | Split on spaces into a list of words | `"buy BTC 10".split()` → `["buy", "BTC", "10"]` |
| `.split(",")` | Split on a specific character | `"a,b,c".split(",")` → `["a", "b", "c"]` |
| `.strip()` | Remove leading/trailing whitespace | `"  hello  ".strip()` → `"hello"` |
| `.upper()` | Convert to uppercase | `"btc".upper()` → `"BTC"` |

Access individual words by index (index starts at 0):

```python
parts = "buy BTC 10".split()
print(parts[0])   # "buy"
print(parts[1])   # "BTC"
print(parts[2])   # "10"   ← still a string; use float() to convert
```

### Handling errors with `try / except`
If code might crash, wrap it in `try` and catch the error type:

```python
try:
    number = int("abc")   # this will fail
except ValueError:
    print("That is not a valid number.")
```

Two common error types for string parsing:
- **`ValueError`** — e.g. `float("abc")` — the text cannot be converted to a number.
- **`IndexError`** — e.g. `parts[4]` when `parts` only has 3 items — the index does not exist.

# Week 5 — Data Structures

**Concepts:** lists · dictionaries · tuples · sets · `sum()` · `max()`

---

## What you need to know

Python has four built-in collection types. Choosing the right one makes your code cleaner and safer.

| Type | Written as | Key property | When to use |
|------|-----------|-------------|-------------|
| **List** | `[a, b, c]` | Ordered, can change | A sequence of items that may grow |
| **Dictionary** | `{"key": value}` | Key-value pairs | Looking something up by name |
| **Tuple** | `(a, b, c)` | Ordered, **cannot change** | A fixed record (e.g. a completed trade) |
| **Set** | `{a, b, c}` | No duplicates | Finding unique values |

### Quick reference

```python
# List — add and access items
fruits = ["apple", "banana"]
fruits.append("cherry")
print(fruits[0])          # "apple"

# Dictionary — look up by key
prices = {"apple": 0.50, "banana": 0.30}
print(prices["apple"])    # 0.50
for item, cost in prices.items():
    print(item, cost)

# Tuple — unpack into named variables
record = ("Alice", 25, "London")
name, age, city = record
print(name)               # "Alice"

# Set — unique values only
visited = {"Paris", "Rome", "Paris"}
print(visited)            # {"Paris", "Rome"}  — no duplicate
```

### Useful built-in functions

```python
numbers = [10, 20, 30]
print(sum(numbers))               # 60
print(max(numbers))               # 30

# max on a dictionary: find the key with the biggest value
stock = {"apple": 50, "banana": 20, "cherry": 80}
top = max(stock, key=stock.get)
print(top)                        # "cherry"
```

**Concepts:** `open()` · write mode `"w"` · read mode `"r"` · `readlines()` · `\n` · `try / except FileNotFoundError`

---

## What you need to know

Right now your program forgets everything when it stops running.  
**File I/O** (Input/Output) lets you save data permanently and reload it next time.

### Writing to a file

```python
with open("myfile.txt", "w") as f:
    f.write("Hello\n")        # \n means "new line"
    f.write("World\n")
```

- `"w"` opens the file for **writing**. If the file already exists it is overwritten.
- `with ... as f:` handles opening and closing the file automatically.
- Every line you write must end with `\n` or the next line will run on from the previous one.

### Reading from a file

```python
with open("myfile.txt", "r") as f:
    lines = f.readlines()      # returns a list — one string per line

for line in lines:
    print(line.strip())        # .strip() removes the trailing \n
```

### Handling a missing file

```python
try:
    with open("data.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("File not found — please create it first.")
```

### CSV format
A **CSV** (Comma-Separated Values) file stores tabular data as plain text:

```
name,age,city
Alice,25,London
Bob,30,Leeds
```

You create it by writing strings with commas between each value:

```python
f.write(f"{name},{age},{city}\n")
```

You read it back by splitting each line on the comma:

```python
parts = line.strip().split(",")
name  = parts[0]
age   = int(parts[1])
city  = parts[2]
```

```python
contacts = [("Alice", "07700900001"), ("Bob", "07700900002")]

    for name, phone in loaded:
```

### Task 6.1 — Save to `trades.csv`

### Task 6.2 — Load from `trades.csv`
Read `trades.csv` back.  
Skip the header row (index 0).  
For each data line, split on commas and rebuild the tuple: `(symbol, action, float(quantity), float(price))`.  
Append it to a list called `loaded_trades` and print how many were loaded.

### Task 6.3 — Handle a missing file

### Task 6.4 — Dashboard
After loading, calculate the total value of all BUY trades and all SELL trades separately

### Expected Output
```
5 trades saved to trades.csv.

```

> **Hint for the dashboard:**  
> Loop through `loaded_trades`. If `action == "BUY"`, add `quantity * price` to a running total.  
> Do the same separately for `"SELL"`.


````python
# Provided data
trades = [
    ("BTC", "BUY",  0.5, 40000),
    ("BTC", "SELL", 0.2, 45000),
    ("ETH", "BUY",  2.0,  2000),
    ("BTC", "BUY",  0.3, 41000),
    ("ETH", "SELL", 1.0,  2200),
]

FILENAME = "trades.csv"

# Step 1: — save trades to FILENAME (write header row then one line per trade)
with open(FILENAME, "w", newline="") as file:
    file.write("symbol,action,quantity,price\n")

    for symbol, action, quantity, price in trades:
        file.write(f"{symbol},{action},{quantity},{price}\n")

print(f"{len(trades)} trades saved to trades.csv.\n")


# Step 2 and 3: load trades from FILENAME into loaded_trades
# wrapping the code in try/ except to find errors and if found, 
# display error message (indexerror or valuerror)

loaded_trades=[]

print(f"Loading {FILENAME}...")

try:
    with open(FILENAME, "r") as file:
     lines = file.readlines()
     for line in lines[1:]:
         parts= line.strip().split(",")
         
         trade_tuple = (parts[0], parts[1], float(parts[2]), float(parts[3]))
         loaded_trades.append(trade_tuple)

    print(f"{len(loaded_trades)} trades loaded.\n")

except FileNotFoundError:
     print(f"Error: The file '{FILENAME}' could not be found.")

# Step 4: [calculate total_bought and total_sold, then print the trade dashboard]
# that is loop through loaded_trades, calculate the total value of bought and sold trades,
# and print the dashboard with total trades, total bought, and total sold.
total_bought = 0.0
total_sold = 0.0

for symbol, action, quantity, price in loaded_trades:
    trade_value = quantity * price
    if action == "BUY":
        total_bought += trade_value
    elif action == "SELL":
        total_sold += trade_value


print("====== TRADE DASHBOARD ======")
print(f"Total trades  : {len(loaded_trades)}")
print(f"Total bought  : £{total_bought:.2f}")
print(f"Total sold    : £{total_sold:.2f}")
print("=============================")

````
